# 3. DBSCAN Clustering – Mall Customers


```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.neighbors import NearestNeighbors
url="https://raw.githubusercontent.com/sharmaroshan/Clustering-of-Mall-Customers/master/Mall_Customers.csv"
df=pd.read_csv(url)
display(df.head())
X=df[["Annual Income (k$)","Spending Score (1-100)"]]
X_scaled=StandardScaler().fit_transform(X)
```

```python
neighbors=NearestNeighbors(n_neighbors=5).fit(X_scaled)
distances,_=neighbors.kneighbors(X_scaled)
k_dist=np.sort(distances[:,-1])
plt.plot(k_dist); plt.xlabel("Sorted points"); plt.ylabel("5-NN distance"); plt.title("K-Distance Graph"); plt.grid(); plt.show()
```

```python
model=DBSCAN(eps=0.50,min_samples=5)
df["Cluster"]=model.fit_predict(X_scaled)
print("Clusters:",sorted(df.Cluster.unique()))
print("Noise points:",(df.Cluster==-1).sum())
display(df.groupby("Cluster")[["Annual Income (k$)","Spending Score (1-100)"]].mean().round(2))
```

```python
plt.scatter(df["Annual Income (k$)"],df["Spending Score (1-100)"],c=df.Cluster,cmap="tab10",s=60)
plt.xlabel("Annual Income (k$)"); plt.ylabel("Spending Score (1-100)"); plt.title("DBSCAN Customer Clusters"); plt.show()
```
