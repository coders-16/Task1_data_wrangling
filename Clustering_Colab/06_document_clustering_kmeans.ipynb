# 6. Document Clustering using TF-IDF + K-Means


```python
!pip -q install scikit-learn
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
import pandas as pd
documents=[
"football match team goal player stadium",
"football player scored a goal in the match",
"cricket match batsman bowler wicket run",
"cricket team won the tournament with many runs",
"python programming code function variable algorithm",
"machine learning model training data prediction",
"python code uses machine learning libraries",
"artificial intelligence machine learning neural network",
"hospital doctor patient treatment medicine",
"doctor examines patient in hospital",
"medicine and healthcare treatment for patients"]
vectorizer=TfidfVectorizer(stop_words="english")
X=vectorizer.fit_transform(documents)
```

```python
k=4
model=KMeans(n_clusters=k,random_state=42,n_init=10)
labels=model.fit_predict(X)
result=pd.DataFrame({"Document":documents,"Cluster":labels})
display(result.sort_values("Cluster"))
```

```python
terms=vectorizer.get_feature_names_out()
order=model.cluster_centers_.argsort()[:,::-1]
for i in range(k):
    print("Cluster",i,":",", ".join(terms[j] for j in order[i,:5]))
```
