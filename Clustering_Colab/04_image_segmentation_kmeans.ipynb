# 4. Image Segmentation using K-Means


```python
!pip -q install scikit-image scikit-learn
import numpy as np, matplotlib.pyplot as plt
from skimage import data
from sklearn.cluster import KMeans
image=data.astronaut()
plt.imshow(image); plt.axis("off"); plt.title("Original Image"); plt.show()
```

```python
pixels=image.reshape(-1,3).astype(float)
k=5
model=KMeans(n_clusters=k,random_state=42,n_init=10)
labels=model.fit_predict(pixels)
segmented=model.cluster_centers_[labels].reshape(image.shape).astype(np.uint8)
plt.imshow(segmented); plt.axis("off"); plt.title(f"Segmented Image - K={k}"); plt.show()
```
